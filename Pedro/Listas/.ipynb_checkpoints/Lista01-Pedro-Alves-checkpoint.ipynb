## Questão 2: Modelo Gaussiano:

### Resposta 2(a)

$$\ell(\mu,\sigma^2) = -\frac{n}{2}\log(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum_{t=1}^n (y_t-\mu)^2$$

Zerando as derivadas, a gente chega em $\hat\mu = \bar y$ e $\hat\sigma^2 = \frac{1}{n}\sum_{t=1}^n (y_t-\bar y)^2$.

```python
mu_hat = y.mean()
sigma2_hat = y.var(ddof=0)   # EMV divide por n
s = y.std(ddof=1)            # s^2 não viesado divide por n-1; usado no intervalo t
t_q = t.ppf(0.975, n - 1)
ci_mu = (mu_hat - t_q * s / np.sqrt(n), mu_hat + t_q * s / np.sqrt(n))

pd.Series({"mu_hat": mu_hat, "sigma2_hat (EMV, /n)": sigma2_hat, "s^2 (/(n-1))": s**2,
           "IC 95% inferior": ci_mu[0], "IC 95% superior": ci_mu[1]}).round(3).to_frame("Modelo Gaussiano")
```

### Resposta 2(b)

$\hat\mu = 30{,}91$, $\hat\sigma^2 = 91{,}59$; IC de 95% para $\mu$: $\bar y \pm t_{0.975,139}\, s/\sqrt n = [29{,}31,\ 32{,}52]$.

```python
grid = np.linspace(0, 65, 400)
fig, ax = plt.subplots()
ax.hist(y, bins=bins, density=True, edgecolor="white", label="dados")
ax.plot(grid, norm.pdf(grid, mu_hat, np.sqrt(sigma2_hat)), lw=2, color="C3",
        label=rf"$N(\hat\mu={mu_hat:.1f},\ \hat\sigma^2={sigma2_hat:.1f})$")
ax.set_xlabel("mortes por mês")
ax.set_ylabel("densidade")
ax.legend()
plt.show()

print(f"P(Y < 0) na normal ajustada = {norm.cdf(0, mu_hat, np.sqrt(sigma2_hat)):.1e}")
```

### Resposta 2(c)

Deixa à desejar, acerta o centro e a dispersão geral, mas erra o formato (os dados reais são assimétricos). A Gaussiana não trabalha que são contagens inteiras e não negativas, que a variância cresce junto com a média e também ignora a dinâmica temporal(sazional).

## Questão 3: Modelo de Poisson:

### Resposta 3(a)

$$\ell(\lambda) = \Big(\sum_t y_t\Big)\log\lambda - n\lambda + \text{constante}, \qquad \ell'(\lambda) = 0 \Rightarrow \hat\lambda = \bar y.$$

$I(\lambda) = n/\lambda$, então o erro padrão é $\text{ep}(\hat\lambda) = \sqrt{\hat\lambda/n}$.

```python
lam_hat = y.mean()
se_lam = np.sqrt(lam_hat / n)
ci_lam = (lam_hat - 1.96 * se_lam, lam_hat + 1.96 * se_lam)

pd.Series({"lambda_hat": lam_hat, "ep = sqrt(lambda_hat/n)": se_lam,
           "IC 95% inferior": ci_lam[0], "IC 95% superior": ci_lam[1]}).round(3).to_frame("Modelo de Poisson")
```

### Resposta 3(b)

$\hat\lambda = 30{,}91$, ep $= 0{,}47$, IC de 95% (Wald) $[29{,}99,\ 31{,}84]$.

```python
obs_bin, _ = np.histogram(y, bins=bins)
exp_pois = n * np.diff(poisson.cdf(np.floor(bins), lam_hat))   # meses esperados por intervalo de largura 3
mids = (bins[:-1] + bins[1:]) / 2
fig, ax = plt.subplots()
ax.bar(mids, obs_bin, width=2.8, color="C0", alpha=0.7, label="observado")
ax.plot(mids, exp_pois, "o-", color="C3", label=rf"esperado, Poisson($\hat\lambda$={lam_hat:.1f})")
ax.set_xlabel("mortes por mês (intervalos de largura 3)")
ax.set_ylabel("quantidade de meses")
ax.legend()
plt.show()
# +- 1 desvio-padrão
pd.DataFrame({
    "observado": [(y <= 20).mean(), ((y > 20) & (y < 42)).mean(), (y >= 42).mean()],
    "Poisson": [poisson.cdf(20, lam_hat), poisson.cdf(41, lam_hat) - poisson.cdf(20, lam_hat),
                poisson.sf(41, lam_hat)],
}, index=["P(y <= 20)", "P(21 <= y <= 41)", "P(y >= 42)"]).round(3)
```

### Resposta 3(c)

A distribuiçao de Poisson fica concentrada mais no intervalo de um desvio-padrão(+-): $[21, 41]$ tem probabilidade 0,942 contra 0,729 observados na prática, e dá só 0,025 e 0,033 para as caldas, quando o observado é 0,136 em cada ponta.

```python
ybar, s2 = y.mean(), y.var(ddof=1)
D = np.sum((y - ybar) ** 2) / ybar   # ~ chi2(n-1) sob o modelo de Poisson
pd.Series({"média": ybar, "variância (ddof=1)": s2, "variância/média": s2 / ybar,
           "estatística de dispersão D": D, "graus de liberdade": n - 1,
           "p-valor P(chi2_139 > D)": chi2.sf(D, n - 1)}).to_frame("teste de dispersão").map(lambda v: f"{v:.4g}")
```

### Resposta 3(d)

Sim, tem uma sobredispersão monstra: variância/média $= 2{,}98$, teste de dispersão $D \approx 3\times10^{-29}$. O erro padrão do modelo de Poisson calculado na 3(b) ficou subestimado.

```python
ll_norm = norm.logpdf(y, mu_hat, np.sqrt(sigma2_hat)).sum()
ll_pois = poisson.logpmf(y, lam_hat).sum()
fit = pd.DataFrame({"k (parâmetros)": [2, 1], "log-verossimilhança máx": [ll_norm, ll_pois]},
                   index=["Gaussiano", "Poisson"])
fit["AIC"] = -2 * fit["log-verossimilhança máx"] + 2 * fit["k (parâmetros)"]
fit.round(2)
```

### Resposta 3(e)

O modelo Gaussiano ganha: AIC de $1033{,}7$ contra $1145{,}0$ da Poisson.

## Questão 4: Série temporal

```python
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(dates, y, marker="o", ms=4, color="C0", label="Mortes mensais")
ax.axhline(mu_hat, color="C3", linestyle="--", label=rf"Média geral ($\bar{{y}}={mu_hat:.1f}$)")
ax.set_xlabel("Data")
ax.set_ylabel("Mortes de motociclistas")
ax.set_title("Série Temporal de Mortes de Motociclistas em SP (2015-2026)")
ax.legend()
plt.show()

# Médias anuais para ilustrar a tendência
df_y = pd.DataFrame({"mortes": y}, index=dates)
print("Média de mortes por ano:")
print(df_y.groupby(df_y.index.year).mean().round(2).T.to_string())
```

### Resposta 4

O gráfico mostra que a hipótese de dados i.i.d não é valida:

1. **Tendência de alta visível:** De 2015 a 2020 a média ficava na casa dos 25 a 27 mortos/mês. De 2021 em diante (pós-pandemia/boom de entregas), a média disparou, passando dos 40 mortos/mês no período recente (2024-2026).
2. **Sazonalidade:** Todo fim de ano (especialmente dezembro) rola um pico isolado de mortes, possivelmente proximo das festas de fim de ano, as pessoas estão se deslocando com mais frequencia, tanto para entrega de aplicativos, como encontros comemorativos.
3. **Autocorrelação:** Um mês alto tende a ser seguido por outro mês alto. Tratar esses dados como observações independentes gera subestimativa das incertezas e ignora a dinâmica do problema.
