### 0.3 Quantities to keep apart

- Test **statistic** ($t=2.47$) vs **p-value** ($0.018$).
- $\log B$ vs $B$ vs $2\log B$. Kass & Raftery categories are defined on $2\log_e B$: 0–2, 2–6, 6–10, >10.
- $\log p(y\mid M)$ is negative, and larger is better ($-122 > -125$).
- **Credible interval** (a probability statement about the parameter, given data and prior) vs Fisher-$z$ **confidence interval** (a frequentist coverage statement).
- **Posterior s.d.** (uncertainty about the parameter) vs **Monte Carlo s.e.** (simulation noise).
- In the t models, $\sigma$ is a **scale**, not the error s.d.: $\operatorname{Var}(\varepsilon)=\sigma^2\nu/(\nu-2)$.
- **Posterior mean of $\beta$** vs **predictive mean of $y_i$**.
- **Pearson** (linear) vs **Spearman** (rank, monotone).
- An outlier in $x$ (**leverage**) vs an outlier in $y$ given $x$ (**residual**).
